```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

path = r"H:\Python\etapa3-pandas\data\Tabela4.csv"

df = pd.read_csv(path, encoding="latin1", sep=";",header=None)

df.dropna()
```

```python

df = df.iloc[1:]

df = df.dropna()
df.reset_index()
```

```python

anos = ['1991', '2000', '2010', '2012', '2013', '2014',
       '2015', '2016', '2017', '2018', '2019', '2020', '2021', '2022', '2023',
       '2024']

df_melt = df.melt(
    id_vars=["Estado"],
    value_vars=anos,
    var_name="Ano",
    value_name="IDHM"
)
```

```python
df_melt
```

```python
df_melt["Ano"] = pd.to_numeric(df_melt["Ano"])
df_melt["IDHM"] = pd.to_numeric(
    df_melt["IDHM"].astype(str).str.replace(",", ".", regex=False),
    errors="coerce"
)
```

```python
df_melt = df_melt.dropna(subset=["IDHM"])
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

for sigla, grupo in df_melt.groupby("Estado"):
    grupo = grupo.sort_values("Ano")
    ax.plot(
        grupo["Ano"],
        grupo["IDHM"],
        marker="o",
        markersize=3,
        linewidth=1.5,
        label=sigla
    )

ax.set_title("Evolução do IDHM por estado")
ax.set_xlabel("Ano")
ax.set_ylabel("IDHM")
ax.set_ylim(0.3, 0.9)
ax.set_xticks(sorted(df_melt["Ano"].unique()))
ax.legend(ncol=3, bbox_to_anchor=(1.02, 1),
          loc="upper left", title="UF")

ax.grid(True, linestyle="--", alpha=0.3)
fig.tight_layout()
plt.show()
```
